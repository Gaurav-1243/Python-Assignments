```python
# Question: Write a Python program to input marks and display the grade:
# 90 to 100: A, 80 to 89: B, 70 to 79: C, Below 70: Fail.

marks = float(input("Enter your marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks!")
elif marks >= 90:
    print("Grade: A")
elif marks >= 80:
    print("Grade: B")
elif marks >= 70:
    print("Grade: C")
else:
    print("Result: Fail")
```

```python
# Question: Write a Python program to verify a user login using a 
# predefined dictionary of email addresses and passwords.

registered_users = {
    "admin@email.com": "Pass123",
    "gaurav@email.com": "Sppu2026"
}

email_input = input("Enter Email: ")
password_input = input("Enter Password: ")

if email_input in registered_users:
    if registered_users[email_input] == password_input:
        print("Login Successful!")
    else:
        print("Error: Incorrect password.")
else:
    print("Error: Email not found.")
```

```python
# Question: Write a Python program to check if a 10-digit mobile number 
# is valid (must be 10 digits, numeric, and start with 6, 7, 8, or 9).

mobile_no = input("Enter your 10-digit mobile number: ")

if len(mobile_no) == 10 and mobile_no.isdigit() and mobile_no[0] in "6789":
    print("Valid Mobile Number.")
else:
    print("Invalid Mobile Number.")

# Regex-based method
import re
mobile_regex = input("Enter mobile number for Regex check: ")
pattern = r"^[6-9]\d{9}$"

if re.match(pattern, mobile_regex):
    print("Valid Mobile Number.")
else:
    print("Invalid Mobile Number.")
```

```python
# Question: Write a Python program to check if a given number 
# is a palindrome (reads the same forward and backward).

# Mathematical logic method
original_num = int(input("Enter a number: "))
temp = original_num
reverse_num = 0

while temp > 0:
    remainder = temp % 10
    reverse_num = (reverse_num * 10) + remainder
    temp = temp // 10

if original_num == reverse_num:
    print("It is a Palindrome")
else:
    print("It is NOT a Palindrome")
```

```python
# s = "aaaaabbbbcccc", output a = 5, b = 4, c = 4

s = "aaaaabbbbcccc"
counts = {}

for i in range(len(s)):
    char = s[i]
    if char in counts:
        counts[char] = counts[char] + 1
    else:
        counts[char] = 1

for char in counts:
    print(char, "=", counts[char])
```

```python
# s1 = "HelloHiBye", output = HHB

s1 = "HelloHiBye"
output = ""

for i in range(len(s1)):
    char = s1[i]
    if char.isupper():
        output = output + char

print(output)
```

```python
# write a program for factorial

n = int(input("Enter a number: "))
factorial = 1

for i in range(1, n + 1):
    factorial = factorial * i

print(factorial)
```

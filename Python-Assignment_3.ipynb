```python
# access element from list by using while loop

l = [32, 23.4, "fsdd", 54, 78, "dhd", "trjy"]
i = 0 

while i < len(l):
    print(l[i])
    i = i + 1  
```

```python
# access element from list by using while loop with decrement

l = [32, 23.4, "fsdd", 54, 78, "dhd", "trjy"]
i = len(l) - 1  

while i >= 0:
    print(l[i])
    i = i - 1   
```

```python
# separate elements into int_list, string_list, and other_list

l = [32, 23.4, "fsdd", 54, 78, "dhd", "trjy"]

int_list = []
string_list = []
other_list = []

for i in range(len(l)):
    item = l[i]
    
    if type(item) == int:
        int_list.append(item)
    elif type(item) == str:
        string_list.append(item)
    else:
        other_list.append(item)

print("int_list =", int_list)
print("string_list =", string_list)
print("other_list =", other_list)
```
